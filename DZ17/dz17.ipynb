Найти рейсы в бизнес-классе, вылетающие из Москвы

```python
select 
	distinct f.flight_no,
	 f.departure_airport,
	 a.airport_name,
	 a.city, 
	 tf.fare_conditions 
from ticket_flights tf 
left join flights f on f.flight_id = tf.flight_id 
left join airports a on a.airport_code = f.departure_airport 
where tf.fare_conditions = 'Business' and f.actual_departure is not null 
	and a.city = 'Москва'
```

--3. Найти самолеты, которые никогда не использовались в рейсах

```python
select 
 	distinct a.model
from aircrafts a 
left join flights f on f.aircraft_code  = a.aircraft_code 
where f.aircraft_code is null
```
