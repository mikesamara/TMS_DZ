```python

create table courses (
	corses_id int primary key,
	course_name varchar(20),
	teacher_name varchar(20),
	departer_name varchar(20),
	departer_phone varchar(20)
	)

create table students(
	student_id int primary key,
	student_name varchar(20)
	)
	
create table grade (
	corses_id int references courses(corses_id),
	student_id int references students(student_id),
	grade int, 
	primary key (corses_id, student_id)
	)
```
