ребования:
promo_id – уникальный ID промокода (первичный ключ, автоинкремент).
code – текст промокода (уникальный, не может повторяться).
discount_percent – процент скидки (от 1 до 100).
valid_from и valid_to – срок действия.
max_uses – максимальное количество применений (NULL = без ограничений).
used_count – сколько раз уже использован (по умолчанию 0).
is_active – активен ли промокод (1 или 0, по умолчанию 1).
created_by – кто создал промокод (внешний ключ на users.id).

```python
create table promocodes (
	promo_id INTEGER PRIMARY KEY AUTOINCREMENT,
	code VARCHAR UNIQUE NOT NULL,
	discont_precent INTEGER CHECK (discont_precent BETWEEN 1 AND 100),
	valid_from DATE NOT NULL,
	valid_to DATE NOT NULL,
	max_uses INTEGER NULL,
	user_count INTEGER DEFAULT 0, 
	is_active INTEGER DEFAULT 1,
	created_by Integer,
	FOREIGN KEY(created_by) REFERENCES users(user_id)
);
```

Базовая статистика по промокодам
Что нужно сделать:
Посчитать:
Общее количество промокодов.
Средний процент скидки .
Количество активных промокодов .

```python
-- кол-во промокодов
SELECT 
 COUNT(*) as cnt_promocodes
FROM 
promocodes;

--средний % скидки
SELECT 
	AVG(discont_precent) as avg_discont_precent
FROM promocodes

-- активные промокоды
SELECT 
	COUNT(*) as activ_promocodes
FROM promocodes
WHERE is_active = 1
```

нализ по группам скидок
Что нужно сделать:
Сгруппировать промокоды по диапазонам скидок и вывести:
Количество промокодов в группе.
Минимальную и максимальную скидку.
Сколько из них имеют ограничение по использованию (max_uses IS NOT NULL).

```python
-- АНАЛИЗ данных
SELECT 
	CASE 
		WHEN discont_precent <= 30 THEN 'до 30%'
		WHEN discont_precent <= 50 THEN 'от 30 до 50%'
		WHEN discont_precent <= 80 THEN 'от 50 до 80%'
		WHEN discont_precent  > 80 THEN 'больше 80%'
		ELSE 'СКИДКИ НЕТ'
	END as group_discont,
	COUNT(*) as cnt_promo, 
	MIN(discont_precent) as min_discont_precent,
	MAX(discont_precent) as max_discont_precent,
	SUM(CASE WHEN max_uses IS NOT NULL THEN 1 ELSE 0 END) as limit_uses
FROM promocodes
GROUP BY group_discont
ORDER BY group_discont
```
