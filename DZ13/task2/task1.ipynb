```python
import requests
import pandas as pd

# Делаем запрос API  по ссылке 
response = requests.get('https://api.spacexdata.com/v3/capsules')
#информацию загружаем в формат json
data = response.json()
# преобразовываем в в формат таблицы
df = pd.DataFrame(data)
# извлекаем и групируем по капсулам каждого типа
cnt_cap = df.groupby('capsule_id').agg(cnt_tp =('capsule_id', 'count'))
print(cnt_cap)

```
