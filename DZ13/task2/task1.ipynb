import requests
import pandas as pd

response = requests.get('https://api.spacexdata.com/v3/capsules')
data = response.json()
df = pd.DataFrame(data)
print(df)
print(df.loc[(df['type'] == 'Dragon 2.0') & (df['status'] == 'active')])

df['date'] = pd.to_datetime(df['original_launch'])
count_df = df.groupby(df['date'].dt.year)['capsule_serial'].count()
print(count_df)
count_df.to_excel('grouped_df.xlsx', index=False)
