Подсчитайте количество транзакций по каждому типу

```python
import pandas as pd
import sqlite3

with sqlite3.connect('database.db') as conn:
    query = 'SELECT * FROM transactions'
    df = pd.read_sql(query, conn)
    cont_type = df.groupby('type').agg(count_type =('transaction_id', 'count'))
    print(cont_type)
```

Определите клиентов с наибольшими расходами (отрицательные суммы транзакций)

```python
import pandas as pd
import sqlite3

with sqlite3.connect('database.db') as conn:
    query = "SELECT * FROM transactions"
    query2 = "SELECT * FROM clients"
    df = pd.read_sql(query, conn)
    df2 = pd.read_sql(query2, conn)
    max_withdramel = df.loc[df['type']=='withdrawal'].groupby('client_id').sum().sort_values('amount').head(3)
    left_join = pd.merge(max_withdramel, df2, on='client_id', how='left')
    print(left_join[['client_id','name', 'amount']])
```

Топ-3 клиента по количеству транзакций

```python
import pandas as pd
import sqlite3

with sqlite3.connect('database.db') as conn:
    query = "SELECT * FROM transactions"
    query2 = 'SELECT * FROM clients'
    df = pd.read_sql(query, conn)
    df2 = pd.read_sql(query2, conn)
    cnt_clint = df.groupby('client_id').agg(cnt_tr=('transaction_id', 'count')).reset_index().sort_values('cnt_tr', ascending=False)
    left_join = pd.merge(cnt_clint, df2, on='client_id', how='left').head(3)
    print(left_join[['client_id', 'name','cnt_tr']])
```
